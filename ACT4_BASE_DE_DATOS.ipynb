```python
%%sql

-- 1. CREACIÓN DE TABLAS PARA SQLITE
CREATE TABLE empleados (
    id_empleado INTEGER PRIMARY KEY AUTOINCREMENT,
    tipo_empleado TEXT CHECK(tipo_empleado IN ('Enfermero Titular', 'Auxiliar de Enfermería', 'Guardia', 'Administrativo')) NOT NULL,
    nombre_completo TEXT NOT NULL,
    direccion TEXT NOT NULL,
    telefono TEXT NOT NULL,
    ciudad TEXT NOT NULL,
    estado TEXT NOT NULL,
    cp TEXT NOT NULL,
    nss TEXT NOT NULL
);

CREATE TABLE medicos (
    id_medico INTEGER PRIMARY KEY AUTOINCREMENT,
    tipo_medico TEXT CHECK(tipo_medico IN ('Titular', 'Interino', 'Sustituto')) NOT NULL,
    nombre_completo TEXT NOT NULL,
    direccion TEXT NOT NULL,
    telefono TEXT NOT NULL,
    ciudad TEXT NOT NULL,
    estado TEXT NOT NULL,
    cp TEXT NOT NULL,
    nss TEXT NOT NULL,
    num_cedula TEXT NOT NULL
);

CREATE TABLE horarios_medicos (
    id_horario INTEGER PRIMARY KEY AUTOINCREMENT,
    id_medico INTEGER NOT NULL,
    dia_semana TEXT CHECK(dia_semana IN ('Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo')) NOT NULL,
    hora_inicio TEXT NOT NULL,
    hora_fin TEXT NOT NULL,
    FOREIGN KEY (id_medico) REFERENCES medicos(id_medico)
);

CREATE TABLE sustituciones (
    id_sustitucion INTEGER PRIMARY KEY AUTOINCREMENT,
    id_medico INTEGER NOT NULL,
    fecha_alta TEXT NOT NULL,
    fecha_baja TEXT NOT NULL,
    FOREIGN KEY (id_medico) REFERENCES medicos(id_medico)
);

CREATE TABLE vacaciones (
    id_vacacion INTEGER PRIMARY KEY AUTOINCREMENT,
    id_persona INTEGER NOT NULL,
    tipo_persona TEXT CHECK(tipo_persona IN ('Medico', 'Empleado')) NOT NULL,
    fecha_inicio TEXT NOT NULL,
    fecha_fin TEXT NOT NULL
);

CREATE TABLE pacientes (
    id_paciente INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre_completo TEXT NOT NULL,
    direccion TEXT NOT NULL,
    telefono TEXT NOT NULL,
    cp TEXT NOT NULL,
    nss TEXT NOT NULL,
    id_medico INTEGER NOT NULL,
    fecha_ingreso TEXT NOT NULL,
    fecha_analisis TEXT,
    fecha_intervencion TEXT,
    fecha_alta TEXT,
    FOREIGN KEY (id_medico) REFERENCES medicos(id_medico)
);


INSERT INTO empleados (tipo_empleado, nombre_completo, direccion, telefono, ciudad, estado, cp, nss) VALUES
('Enfermero Titular', 'Ana Rodríguez', 'Calle Pino 123', '8110001001', 'Monterrey', 'Nuevo León', '64000', 'NSS1000001'),
('Auxiliar de Enfermería', 'Luis Garza', 'Av. Madero 456', '8110001002', 'Guadalupe', 'Nuevo León', '67100', 'NSS1000002'),
('Guardia', 'Carlos Torres', 'Calle Loma 789', '8110001003', 'San Pedro', 'Nuevo León', '66200', 'NSS1000003'),
('Administrativo', 'Diana Silva', 'Blvd. Diaz 101', '8110001004', 'Apodaca', 'Nuevo León', '66600', 'NSS1000004'),
('Enfermero Titular', 'Eva Luna', 'Calle Roble 202', '8110001005', 'San Nicolás', 'Nuevo León', '66400', 'NSS1000005'),
('Auxiliar de Enfermería', 'Fernando Rey', 'Av. Central 303', '8110001006', 'Monterrey', 'Nuevo León', '64010', 'NSS1000006'),
('Guardia', 'Gilberto Paz', 'Calle Real 404', '8110001007', 'Guadalupe', 'Nuevo León', '67110', 'NSS1000007'),
('Administrativo', 'Hilda Gomez', 'Av. Sur 505', '8110001008', 'Escobedo', 'Nuevo León', '66050', 'NSS1000008'),
('Enfermero Titular', 'Ignacio Soto', 'Calle Norte 606', '8110001009', 'Santa Catarina', 'Nuevo León', '66100', 'NSS1000009'),
('Auxiliar de Enfermería', 'Julia Rios', 'Av. Universidad 707', '8110001010', 'San Nicolás', 'Nuevo León', '66450', 'NSS1000010'),
('Guardia', 'Karim Ortiz', 'Calle Mina 808', '8110001011', 'Monterrey', 'Nuevo León', '64020', 'NSS1000011'),
('Administrativo', 'Laura Cruz', 'Blvd. Acero 909', '8110001012', 'Apodaca', 'Nuevo León', '66610', 'NSS1000012'),
('Enfermero Titular', 'Mario Ruiz', 'Calle Valle 111', '8110001013', 'San Pedro', 'Nuevo León', '66210', 'NSS1000013'),
('Auxiliar de Enfermería', 'Nora Vera', 'Av. Hidalgo 222', '8110001014', 'Guadalupe', 'Nuevo León', '67120', 'NSS1000014'),
('Guardia', 'Omar Lara', 'Calle Sol 333', '8110001015', 'Monterrey', 'Nuevo León', '64030', 'NSS1000015'),
('Administrativo', 'Paula Mier', 'Blvd. Constitución 444', '8110001016', 'Monterrey', 'Nuevo León', '64040', 'NSS1000016'),
('Enfermero Titular', 'Quirino Gil', 'Calle Luna 555', '8110001017', 'San Nicolás', 'Nuevo León', '66460', 'NSS1000017'),
('Auxiliar de Enfermería', 'Rosa Mora', 'Av. Las Palmas 666', '8110001018', 'Escobedo', 'Nuevo León', '66060', 'NSS1000018'),
('Guardia', 'Saul Vega', 'Calle Estrella 777', '8110001019', 'Apodaca', 'Nuevo León', '66620', 'NSS1000019'),
('Administrativo', 'Tania Roa', 'Blvd. Rey 888', '8110001020', 'Santa Catarina', 'Nuevo León', '66110', 'NSS1000020'),
('Enfermero Titular', 'Ulises cano', 'Calle Roca 999', '8110001021', 'Monterrey', 'Nuevo León', '64050', 'NSS1000021'),
('Auxiliar de Enfermería', 'Vera Rios', 'Av. Lago 121', '8110001022', 'San Pedro', 'Nuevo León', '66220', 'NSS1000022'),
('Guardia', 'Walter Pi', 'Calle Rio 232', '8110001023', 'Guadalupe', 'Nuevo León', '67130', 'NSS1000023'),
('Administrativo', 'Ximena luz', 'Blvd. Mar 343', '8110001024', 'San Nicolás', 'Nuevo León', '66470', 'NSS1000024'),
('Enfermero Titular', 'Yahir Solis', 'Calle Brisa 454', '8110001025', 'Monterrey', 'Nuevo León', '64060', 'NSS1000025'),
('Auxiliar de Enfermería', 'Zoe Rivas', 'Av. Viento 565', '8110001026', 'Apodaca', 'Nuevo León', '66630', 'NSS1000026'),
('Guardia', 'Abel Cota', 'Calle Nieve 676', '8110001027', 'Escobedo', 'Nuevo León', '66070', 'NSS1000027'),
('Administrativo', 'Berta Peña', 'Blvd. Fuego 787', '8110001028', 'Santa Catarina', 'Nuevo León', '66120', 'NSS1000028'),
('Enfermero Titular', 'Cesar Oca', 'Calle Bosque 898', '8110001029', 'Monterrey', 'Nuevo León', '64070', 'NSS1000029'),
('Auxiliar de Enfermería', 'Dalia Cien', 'Av. Selva 909', '8110001030', 'San Pedro', 'Nuevo León', '66230', 'NSS1000030');

INSERT INTO medicos (tipo_medico, nombre_completo, direccion, telefono, ciudad, estado, cp, nss, num_cedula) VALUES
-- 1 a 10 Titulares
('Titular', 'Dr. Alfonso Reyes', 'Calle Plata 10', '8120002001', 'Monterrey', 'Nuevo León', '64000', 'NSS2001', 'CED001'),
('Titular', 'Dra. Blanca Treviño', 'Av. Oro 20', '8120002002', 'San Pedro', 'Nuevo León', '66200', 'NSS2002', 'CED002'),
('Titular', 'Dr. Carlos Slim', 'Calle Cobre 30', '8120002003', 'Guadalupe', 'Nuevo León', '67100', 'NSS2003', 'CED003'),
('Titular', 'Dra. Daniela Romo', 'Blvd. Zinc 40', '8120002004', 'Apodaca', 'Nuevo León', '66600', 'NSS2004', 'CED004'),
('Titular', 'Dr. Emilio Azcarraga', 'Calle Niquel 50', '8120002005', 'San Nicolás', 'Nuevo León', '66400', 'NSS2005', 'CED005'),
('Titular', 'Dra. Flor Silvestre', 'Av. Plomo 60', '8120002006', 'Monterrey', 'Nuevo León', '64010', 'NSS2006', 'CED006'),
('Titular', 'Dr. Gabriel Garcia', 'Calle Litio 70', '8120002007', 'Escobedo', 'Nuevo León', '66050', 'NSS2007', 'CED007'),
('Titular', 'Dra. Helena Rojo', 'Blvd. Sodio 80', '8120002008', 'Santa Catarina', 'Nuevo León', '66100', 'NSS2008', 'CED008'),
('Titular', 'Dr. Ismael Isaza', 'Calle Radio 90', '8120002009', 'Monterrey', 'Nuevo León', '64020', 'NSS2009', 'CED009'),
('Titular', 'Dra. Juana Ines', 'Av. Titanio 100', '8120002010', 'San Pedro', 'Nuevo León', '66210', 'NSS2010', 'CED010'),
-- 11 a 20 Interinos (El 11 será clave para la Q7 y Q1-Q3)
('Interino', 'Dr. Kevin Mier', 'Calle Hierro 110', '8120002011', 'Guadalupe', 'Nuevo León', '67110', 'NSS2011', 'CED011'),
('Interino', 'Dra. Leticia Ortiz', 'Blvd. Cromo 120', '8120002012', 'Apodaca', 'Nuevo León', '66610', 'NSS2012', 'CED012'),
('Interino', 'Dr. Mario Vargas', 'Calle Bario 130', '8120002013', 'San Nicolás', 'Nuevo León', '66410', 'NSS2013', 'CED013'),
('Interino', 'Dra. Nadia Ali', 'Av. Cesio 140', '8120002014', 'Monterrey', 'Nuevo León', '64030', 'NSS2014', 'CED014'),
('Interino', 'Dr. Oscar Wilde', 'Calle Neon 150', '8120002015', 'Escobedo', 'Nuevo León', '66060', 'NSS2015', 'CED015'),
('Interino', 'Dra. Perla Silva', 'Blvd. Argon 160', '8120002016', 'Santa Catarina', 'Nuevo León', '66110', 'NSS2016', 'CED016'),
('Interino', 'Dr. Quinton Jax', 'Calle Xenon 170', '8120002017', 'Monterrey', 'Nuevo León', '64040', 'NSS2017', 'CED017'),
('Interino', 'Dra. Raquel Diaz', 'Av. Radon 180', '8120002018', 'San Pedro', 'Nuevo León', '66220', 'NSS2018', 'CED018'),
('Interino', 'Dr. Sergio Ramos', 'Calle Cobalto 190', '8120002019', 'Guadalupe', 'Nuevo León', '67120', 'NSS2019', 'CED019'),
('Interino', 'Dra. Teresa Oca', 'Blvd. Uranio 200', '8120002020', 'Apodaca', 'Nuevo León', '66620', 'NSS2020', 'CED020'),
-- 21 a 30 Sustitutos (El 21 para Q1-Q4, el 22 para Q5-Q6)
('Sustituto', 'Dr. Uriel Antuna', 'Calle Plutonio 210', '8120002021', 'San Nicolás', 'Nuevo León', '66420', 'NSS2021', 'CED021'),
('Sustituto', 'Dra. Valeria Marin', 'Av. Galio 220', '8120002022', 'Monterrey', 'Nuevo León', '64050', 'NSS2022', 'CED022'),
('Sustituto', 'Dr. Walter White', 'Calle Indio 230', '8120002023', 'Escobedo', 'Nuevo León', '66070', 'NSS2023', 'CED023'),
('Sustituto', 'Dra. Xochitl Gomez', 'Blvd. Talio 240', '8120002024', 'Santa Catarina', 'Nuevo León', '66120', 'NSS2024', 'CED024'),
('Sustituto', 'Dr. Yandel Rios', 'Calle Mercurio 250', '8120002025', 'Monterrey', 'Nuevo León', '64060', 'NSS2025', 'CED025'),
('Sustituto', 'Dra. Zuria Vega', 'Av. Bismuto 260', '8120002026', 'San Pedro', 'Nuevo León', '66230', 'NSS2026', 'CED026'),
('Sustituto', 'Dr. Aldo De Nigris', 'Calle Cobre 270', '8120002027', 'Guadalupe', 'Nuevo León', '67130', 'NSS2027', 'CED027'),
('Sustituto', 'Dra. Bety Monroe', 'Blvd. Zinc 280', '8120002028', 'Apodaca', 'Nuevo León', '66630', 'NSS2028', 'CED028'),
('Sustituto', 'Dr. Cristian Castro', 'Calle Plata 290', '8120002029', 'San Nicolás', 'Nuevo León', '66430', 'NSS2029', 'CED029'),
('Sustituto', 'Dra. Danna Paola', 'Av. Oro 300', '8120002030', 'Monterrey', 'Nuevo León', '64070', 'NSS2030', 'CED030');

-- Altas y bajas (Para sustitutos)
INSERT INTO sustituciones (id_medico, fecha_alta, fecha_baja) VALUES
(21, '2023-01-01', '2026-12-31'), (22, '2023-01-01', '2026-12-31');

-- Vacaciones estratégicas para que las consultas arrojen resultados precisos
INSERT INTO vacaciones (id_persona, tipo_persona, fecha_inicio, fecha_fin) VALUES
-- Medico 21 (Sustituto) - 3 periodos en 2024 (< Julio). Uno de < 3 días. Meses Feb, Abr, Jun (Cumple Q1, Q2, Q3, Q4)
(21, 'Medico', '2024-02-10', '2024-02-11'), -- 2 días (Febrero)
(21, 'Medico', '2024-04-10', '2024-04-14'), -- 5 días (Abril)
(21, 'Medico', '2024-06-10', '2024-06-13'), -- 4 días (Junio) Total 11 días.

-- Medico 11 (Interino) - 3 periodos en 2024 (< Julio). Uno < 3 días. (Cumple Q1, Q2, Q3. No Q4 por no ser sustituto)
(11, 'Medico', '2024-03-01', '2024-03-02'),
(11, 'Medico', '2024-05-01', '2024-05-05'),
(11, 'Medico', '2024-06-01', '2024-06-05'),

-- Medico 22 (Sustituto) - Vacaciones Feb-Mar 2025 para Q5, Q6
(22, 'Medico', '2025-02-15', '2025-02-25'),

-- Medico 12 (Interino) - Vacaciones Feb-Mar 2025 para Q7
(12, 'Medico', '2025-03-10', '2025-03-20');

INSERT INTO pacientes (nombre_completo, direccion, telefono, cp, nss, id_medico, fecha_ingreso, fecha_analisis, fecha_intervencion, fecha_alta) VALUES
('Paciente 01', 'Calle A 1', '8130003001', '64000', 'PAC01', 1, '2024-01-01', NULL, NULL, '2024-01-10'),
('Paciente 02', 'Calle B 2', '8130003002', '64010', 'PAC02', 2, '2024-01-02', '2024-01-03', '2024-01-05', '2024-01-15'),
('Paciente 03', 'Calle C 3', '8130003003', '64020', 'PAC03', 3, '2024-02-01', NULL, NULL, '2024-02-10'),
('Paciente 04', 'Calle D 4', '8130003004', '64030', 'PAC04', 4, '2024-03-01', '2024-03-02', '2024-03-03', '2024-03-10'),
('Paciente 05', 'Calle E 5', '8130003005', '64040', 'PAC05', 5, '2024-04-01', NULL, NULL, '2024-04-10'),
('Paciente 06', 'Calle F 6', '8130003006', '64050', 'PAC06', 6, '2024-05-01', '2024-05-02', '2024-05-05', '2024-05-15'),
('Paciente 07', 'Calle G 7', '8130003007', '64060', 'PAC07', 7, '2024-06-01', NULL, NULL, '2024-06-10'),
('Paciente 08', 'Calle H 8', '8130003008', '64070', 'PAC08', 8, '2024-07-01', '2024-07-02', '2024-07-03', '2024-07-10'),
('Paciente 09', 'Calle I 9', '8130003009', '64080', 'PAC09', 9, '2024-08-01', NULL, NULL, '2024-08-10'),
('Paciente 10', 'Calle J 10', '8130003010', '64090', 'PAC10', 10, '2024-09-01', '2024-09-02', '2024-09-05', '2024-09-15'),
('Paciente 11', 'Calle K 11', '8130003011', '64100', 'PAC11', 11, '2024-10-01', NULL, NULL, '2024-10-10'),
('Paciente 12', 'Calle L 12', '8130003012', '64110', 'PAC12', 12, '2025-03-01', '2025-03-15', NULL, '2025-03-20'), -- Q7: Análisis 2025-03-15 empata con vacación
('Paciente 13', 'Calle M 13', '8130003013', '64120', 'PAC13', 13, '2024-12-01', NULL, NULL, '2024-12-10'),
('Paciente 14', 'Calle N 14', '8130003014', '64130', 'PAC14', 14, '2025-01-01', '2025-01-02', '2025-01-05', '2025-01-15'),
('Paciente 15', 'Calle O 15', '8130003015', '64140', 'PAC15', 15, '2025-02-01', NULL, NULL, '2025-02-10'),
('Paciente 16', 'Calle P 16', '8130003016', '64150', 'PAC16', 16, '2025-03-01', '2025-03-02', '2025-03-05', '2025-03-15'),
('Paciente 17', 'Calle Q 17', '8130003017', '64160', 'PAC17', 17, '2025-04-01', NULL, NULL, '2025-04-10'),
('Paciente 18', 'Calle R 18', '8130003018', '64170', 'PAC18', 18, '2025-05-01', '2025-05-02', '2025-05-05', '2025-05-15'),
('Paciente 19', 'Calle S 19', '8130003019', '64180', 'PAC19', 19, '2025-06-01', NULL, NULL, '2025-06-10'),
('Paciente 20', 'Calle T 20', '8130003020', '64190', 'PAC20', 20, '2025-07-01', '2025-07-02', '2025-07-05', '2025-07-15'),
('Paciente 21', 'Calle U 21', '8130003021', '64200', 'PAC21', 21, '2024-03-01', NULL, NULL, '2024-03-10'),
('Paciente 22', 'Calle V 22', '8130003022', '64210', 'PAC22', 22, '2025-02-10', '2025-02-12', '2025-02-20', '2025-02-28'), -- Q5 y Q6: Intervención en vacación
('Paciente 23', 'Calle W 23', '8130003023', '64220', 'PAC23', 23, '2025-03-01', NULL, NULL, '2025-03-10'),
('Paciente 24', 'Calle X 24', '8130003024', '64230', 'PAC24', 24, '2025-04-01', '2025-04-02', '2025-04-05', '2025-04-15'),
('Paciente 25', 'Calle Y 25', '8130003025', '64240', 'PAC25', 25, '2025-05-01', NULL, NULL, '2025-05-10'),
('Paciente 26', 'Calle Z 26', '8130003026', '64250', 'PAC26', 26, '2025-06-01', '2025-06-02', '2025-06-05', '2025-06-15'),
('Paciente 27', 'Calle AA 27', '8130003027', '64260', 'PAC27', 27, '2025-07-01', NULL, NULL, '2025-07-10'),
('Paciente 28', 'Calle BB 28', '8130003028', '64270', 'PAC28', 28, '2025-08-01', '2025-08-02', '2025-08-05', '2025-08-15'),
('Paciente 29', 'Calle CC 29', '8130003029', '64280', 'PAC29', 29, '2025-09-01', NULL, NULL, '2025-09-10'),
('Paciente 30', 'Calle DD 30', '8130003030', '64290', 'PAC30', 30, '2025-10-01', '2025-10-02', '2025-10-05', '2025-10-15');
```

```python
%%sql
SELECT DISTINCT m.nombre_completo
FROM medicos m
JOIN vacaciones v ON m.id_medico = v.id_persona AND v.tipo_persona = 'Medico'
WHERE strftime('%Y', v.fecha_inicio) = '2024'
  AND CAST(strftime('%m', v.fecha_inicio) AS INTEGER) < 7
ORDER BY m.nombre_completo ASC;
```

```python
%%sql
SELECT m.nombre_completo
FROM medicos m
JOIN vacaciones v ON m.id_medico = v.id_persona AND v.tipo_persona = 'Medico'
WHERE m.tipo_medico IN ('Interino', 'Sustituto')
  AND strftime('%Y', v.fecha_inicio) = '2024'
  AND CAST(strftime('%m', v.fecha_inicio) AS INTEGER) < 7
GROUP BY m.id_medico, m.nombre_completo
HAVING COUNT(v.id_vacacion) = 3;
```

```python
%%sql
SELECT m.nombre_completo
FROM medicos m
JOIN vacaciones v ON m.id_medico = v.id_persona AND v.tipo_persona = 'Medico'
WHERE m.tipo_medico IN ('Interino', 'Sustituto')
  AND strftime('%Y', v.fecha_inicio) = '2024'
  AND CAST(strftime('%m', v.fecha_inicio) AS INTEGER) < 7
GROUP BY m.id_medico, m.nombre_completo
HAVING COUNT(v.id_vacacion) = 3
   -- Comprobamos que la diferencia de días entre inicio y fin (más 1) sea menor a 3 días
   AND SUM(CASE WHEN (julianday(v.fecha_fin) - julianday(v.fecha_inicio)) + 1 < 3 THEN 1 ELSE 0 END) > 0;
```

```python
%%sql
SELECT m.nombre_completo
FROM medicos m
JOIN vacaciones v ON m.id_medico = v.id_persona AND v.tipo_persona = 'Medico'
WHERE m.tipo_medico = 'Sustituto'
  AND strftime('%Y', v.fecha_inicio) = '2024'
  AND CAST(strftime('%m', v.fecha_inicio) AS INTEGER) < 7
GROUP BY m.id_medico, m.nombre_completo
HAVING COUNT(v.id_vacacion) = 3
   AND SUM(CASE WHEN (julianday(v.fecha_fin) - julianday(v.fecha_inicio)) + 1 < 3 THEN 1 ELSE 0 END) > 0
   AND SUM(CASE WHEN CAST(strftime('%m', v.fecha_inicio) AS INTEGER) = 2 THEN 1 ELSE 0 END) > 0
   AND SUM(CASE WHEN CAST(strftime('%m', v.fecha_inicio) AS INTEGER) = 4 THEN 1 ELSE 0 END) > 0
   AND SUM(CASE WHEN CAST(strftime('%m', v.fecha_inicio) AS INTEGER) = 6 THEN 1 ELSE 0 END) > 0;
```

```python
%%sql
SELECT m.nombre_completo AS Medico,
       v.fecha_inicio || ' a ' || v.fecha_fin AS Periodo_Vacacional,
       p.nombre_completo AS Paciente
FROM medicos m
JOIN vacaciones v ON m.id_medico = v.id_persona AND v.tipo_persona = 'Medico'
JOIN pacientes p ON p.id_medico = m.id_medico
WHERE strftime('%Y', v.fecha_inicio) = '2025'
  AND CAST(strftime('%m', v.fecha_inicio) AS INTEGER) IN (2, 3)
  AND p.fecha_intervencion >= v.fecha_inicio
  AND p.fecha_intervencion <= v.fecha_fin;
```

```python
%%sql
SELECT m.nombre_completo AS Medico_Sustituto,
       v.fecha_inicio || ' a ' || v.fecha_fin AS Periodo_Vacacional,
       p.nombre_completo AS Paciente,
       p.fecha_intervencion
FROM medicos m
JOIN vacaciones v ON m.id_medico = v.id_persona AND v.tipo_persona = 'Medico'
JOIN pacientes p ON p.id_medico = m.id_medico
WHERE m.tipo_medico = 'Sustituto'
  AND strftime('%Y', v.fecha_inicio) = '2025'
  AND CAST(strftime('%m', v.fecha_inicio) AS INTEGER) IN (2, 3)
  AND p.fecha_intervencion >= v.fecha_inicio
  AND p.fecha_intervencion <= v.fecha_fin
ORDER BY p.fecha_intervencion ASC;
```

```python
%%sql
SELECT m.nombre_completo AS Medico_Interino,
       v.fecha_inicio || ' a ' || v.fecha_fin AS Periodo_Vacacional,
       p.nombre_completo AS Paciente,
       p.fecha_analisis
FROM medicos m
JOIN vacaciones v ON m.id_medico = v.id_persona AND v.tipo_persona = 'Medico'
JOIN pacientes p ON p.id_medico = m.id_medico
WHERE m.tipo_medico = 'Interino'
  AND p.fecha_analisis >= v.fecha_inicio
  AND p.fecha_analisis <= v.fecha_fin;
```
